```python
# TASK 1: 

def minimax_robot():
    
    tree = {
        "Zone A": [3, 5],
        "Zone B": [2, 9]
    }

    print("Game Tree:")
    for zone in tree:
        print(f"{zone} -> {tree[zone]}")

    best_value = float('-inf')
    best_move = None

    print("\nEvaluating using Minimax:\n")

    for zone in tree:
        print(f"Robot chooses {zone}")

       
        min_value = min(tree[zone])
        print(f"Thief chooses minimum from {tree[zone]} = {min_value}")

        if min_value > best_value:
            best_value = min_value
            best_move = zone

        print(f"Value for {zone} = {min_value}\n")

    print("Final Decision:")
    print(f"Robot should choose {best_move} with value {best_value}")



minimax_robot()
```

```python
# TASK 2


tree = {
    "Stop": [6, 7],
    "Go": [2, 4],
    "Turn": [8, 3]
}


def minimax_car():
    best_value = float('-inf')
    best_move = None

    print("\n--- MINIMAX ---\n")

    for move in tree:
        print(f"Car chooses {move}")

        
        value = min(tree[move])
        print(f"Opponent chooses min {tree[move]} = {value}")

        if value > best_value:
            best_value = value
            best_move = move

        print(f"Value of {move} = {value}\n")

    print(f"Best Move: {best_move} with value {best_value}")



def alpha_beta_car():
    print("\n--- ALPHA-BETA PRUNING ---\n")

    alpha = float('-inf')
    beta = float('inf')

    best_move = None
    nodes_evaluated = 0

    for move in tree:
        print(f"\nEvaluating {move}")

        value = float('inf')

        for outcome in tree[move]:
            nodes_evaluated += 1
            value = min(value, outcome)

            print(f"Checking value {outcome}")
            beta = min(beta, value)

            if beta <= alpha:
                print("Pruned remaining branch!")
                break

        print(f"Value for {move} = {value}")

        if value > alpha:
            alpha = value
            best_move = move

    print("\nFinal Decision:")
    print(f"Best Move: {best_move}")
    print(f"Nodes evaluated: {nodes_evaluated}")



minimax_car()
alpha_beta_car()
```

```python
# TASK 3


tree = {
    "Attack": {
        "A1": [5, 6],
        "A2": [7, 4]
    },
    "Defend": {
        "D1": [3, 8],
        "D2": [6, 2]
    },
    "Gather": {
        "G1": [1, 9],
        "G2": [4, 7]
    }
}


def minimax_game():
    print("\n--- MINIMAX ---\n")

    best_value = float('-inf')
    best_move = None
    nodes = 0

    for action in tree:
        print(f"\nEvaluating {action}")

        min_value = float('inf')

        for sub_action in tree[action]:
            values = tree[action][sub_action]
            max_value = max(values)
            nodes += 2

            print(f"{sub_action} -> max{values} = {max_value}")

            min_value = min(min_value, max_value)

        print(f"{action} value = {min_value}")

        if min_value > best_value:
            best_value = min_value
            best_move = action

    print("\nFinal Decision:")
    print(f"Best Action: {best_move}")
    print(f"Nodes evaluated: {nodes}")



def alpha_beta_game():
    print("\n--- ALPHA-BETA PRUNING ---\n")

    alpha = float('-inf')
    beta = float('inf')
    nodes = 0

    best_move = None

    for action in tree:
        print(f"\nAction: {action}")
        value = float('inf')

        for sub_action in tree[action]:
            max_value = float('-inf')

            for v in tree[action][sub_action]:
                nodes += 1
                max_value = max(max_value, v)

                print(f"Checking {v} (α={alpha}, β={beta})")

                if max_value >= beta:
                    print("Pruned!")
                    break

            value = min(value, max_value)
            beta = min(beta, value)

            if beta <= alpha:
                print("Pruned subtree!")
                break

        print(f"{action} value = {value}")

        if value > alpha:
            alpha = value
            best_move = action

    print("\nFinal Decision:")
    print(f"Best Action: {best_move}")
    print(f"Nodes evaluated: {nodes}")



minimax_game()
alpha_beta_game()
```
