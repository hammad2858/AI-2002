```python
#task2
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

model = DiscreteBayesianNetwork([
    ('Intelligence', 'Grade'),
    ('StudyHours', 'Grade'),
    ('Difficulty', 'Grade'),
    ('Grade', 'Pass')
])


cpd_I = TabularCPD(variable='Intelligence', variable_card=2,
                   values=[[0.7], [0.3]])

cpd_S = TabularCPD(variable='StudyHours', variable_card=2,
                   values=[[0.6], [0.4]])

cpd_D = TabularCPD(variable='Difficulty', variable_card=2,
                   values=[[0.4], [0.6]])


cpd_G = TabularCPD(
    variable='Grade', variable_card=3,
    values=[
        # A
        [0.9, 0.7, 0.6, 0.3, 0.7, 0.5, 0.4, 0.2],
        # B
        [0.08, 0.2, 0.3, 0.4, 0.2, 0.3, 0.4, 0.5],
        # C
        [0.02, 0.1, 0.1, 0.3, 0.1, 0.2, 0.2, 0.3],
    ],
    evidence=['Intelligence', 'StudyHours', 'Difficulty'],
    evidence_card=[2, 2, 2]
)


cpd_P = TabularCPD(
    variable='Pass', variable_card=2,
    values=[
        [0.95, 0.80, 0.50],  
        [0.05, 0.20, 0.50]   
    ],
    evidence=['Grade'],
    evidence_card=[3]
)


model.add_cpds(cpd_I, cpd_S, cpd_D, cpd_G, cpd_P)


print("Model valid:", model.check_model())


infer = VariableElimination(model)


q1 = infer.query(variables=['Pass'],
                 evidence={'StudyHours': 0, 'Difficulty': 0})
print("\nP(Pass | Sufficient, Hard):")
print(q1)


q2 = infer.query(variables=['Intelligence'],
                 evidence={'Pass': 0})
print("\nP(Intelligence | Pass=Yes):")
print(q2)
```

```python
#task3 
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD   
from pgmpy.inference import VariableElimination

model = DiscreteBayesianNetwork([
    ('Disease', 'Fever'),
    ('Disease', 'Cough'),
    ('Disease', 'Fatigue'),
    ('Disease', 'Chills')
])
cpd_Disease = TabularCPD(variable='Disease', variable_card=2,
                         values=[[0.3], [0.7]])
cpd_Fever = TabularCPD(
    variable='Fever', variable_card=2,
    values=[[0.9, 0.5], [0.1, 0.5]],
    evidence=['Disease'], evidence_card=[2]
)

cpd_cough = TabularCPD(
    variable='Cough', variable_card=2,
    values=[[0.8, 0.6], [0.2, 0.4]],
    evidence=['Disease'], evidence_card=[2]
)

cpd_fatigue = TabularCPD(
    variable='Fatigue', variable_card=2,
    values=[[0.7, 0.3], [0.3, 0.7]],
    evidence=['Disease'], evidence_card=[2]
)
cpd_chills = TabularCPD(
    variable='Chills', variable_card=2,
    values=[[0.6, 0.4], [0.4, 0.6]],
    evidence=['Disease'], evidence_card=[2]
)


model.add_cpds(cpd_Disease,cpd_Fever,cpd_cough,cpd_fatigue,cpd_chills)

print("Model valid:", model.check_model())
infer = VariableElimination(model)
q1 = infer.query(variables=['Disease'],
                 evidence={'Fever': 0, 'Cough': 0})
print("\nP(Disease | Fever, Cough):")
print(q1)


q2 = infer.query(variables=['Disease'],
                 evidence={'Fever': 0, 'Cough': 0, 'Chills': 0})
print("\nP(Disease | Fever, Cough, Chills):")
print(q2)


q3 = infer.query(variables=['Fatigue'],
                 evidence={'Disease': 0})
print("\nP(Fatigue | Flu):")
print(q3)
```

```python
#task4
import numpy as np
import random 
states = ['Sunny', 'Rainy', 'Cloudy']

transition_matrix = np.array([
    [0.6, 0.3, 0.1],
    [0.2,0.5,0.3],
    [0.2,0.3,0.5]
])

states_index = {state: index for index, state in enumerate(states)}
def simulate_weather(days,start='Sunny'):
    current = start
    sequence = [current]
    for _ in range(days-1):
        probs = transition_matrix[states_index[current]]
        current = random.choices(states, probs)[0]
        sequence.append(current)
    return sequence

weather_seq = simulate_weather(10)
print("Weather for 10 days :", weather_seq)
rainy_days = weather_seq.count('Rainy')
print("Rainy days:", rainy_days)

trials = 10000
counts =0

for _ in range(trials):
    if simulate_weather(3)[-1] == 'Rainy':
        counts += 1

print("Estimated probability of Rainy day after 3 days:", counts/trials)
```
